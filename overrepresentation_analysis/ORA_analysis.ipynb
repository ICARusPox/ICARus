```python
# Load the extended ORA results tables
# Unrefined intensities
early_low_unrefined_int_ora_extended_path = (
    "DAVID_Func_Annot_Chart_ORA_"
    "results_unrefined_ints/DAVIDChartReport_top_250_genes_early_low_"
    "unrefined_int_2026-08-03_extended.csv"
)
early_low_unrefined_int_ora_extended_df = pd.read_csv(
    early_low_unrefined_int_ora_extended_path
)

early_high_unrefined_int_ora_extended_path = (
    "DAVID_Func_Annot_Chart_ORA_"
    "results_unrefined_ints/DAVIDChartReport_top_250_genes_early_high_"
    "unrefined_int_2026-08-03_extended.csv"
)
early_high_unrefined_int_ora_extended_df = pd.read_csv(
    early_high_unrefined_int_ora_extended_path
)

late_low_unrefined_int_ora_extended_path = (
    "DAVID_Func_Annot_Chart_ORA_"
    "results_unrefined_ints/DAVIDChartReport_top_250_genes_late_low_"
    "unrefined_int_2026-08-03_extended.csv"
)
late_low_unrefined_int_ora_extended_df = pd.read_csv(
    late_low_unrefined_int_ora_extended_path
)

late_high_unrefined_int_ora_extended_path = (
    "DAVID_Func_Annot_Chart_ORA_"
    "results_unrefined_ints/DAVIDChartReport_top_250_genes_late_high_"
    "unrefined_int_2026-08-03_extended.csv"
)
late_high_unrefined_int_ora_extended_df = pd.read_csv(
    late_high_unrefined_int_ora_extended_path
)

# Refined intensities
early_low_refined_int_ora_extended_path = (
    "DAVID_Func_Annot_Chart_ORA_"
    "results_refined_ints/DAVIDChartReport_top_250_genes_early_low_"
    "refined_int_2026-08-03_extended.csv"
)
early_low_refined_int_ora_extended_df = pd.read_csv(
    early_low_refined_int_ora_extended_path
)

early_high_refined_int_ora_extended_path = (
    "DAVID_Func_Annot_Chart_ORA_"
    "results_refined_ints/DAVIDChartReport_top_250_genes_early_high_"
    "refined_int_2026-08-03_extended.csv"
)
early_high_refined_int_ora_extended_df = pd.read_csv(
    early_high_refined_int_ora_extended_path
)

late_low_refined_int_ora_extended_path = (
    "DAVID_Func_Annot_Chart_ORA_"
    "results_refined_ints/DAVIDChartReport_top_250_genes_late_low_"
    "refined_int_2026-08-03_extended.csv"
)
late_low_refined_int_ora_extended_df = pd.read_csv(
    late_low_refined_int_ora_extended_path
)

late_high_refined_int_ora_extended_path = (
    "DAVID_Func_Annot_Chart_ORA_"
    "results_refined_ints/DAVIDChartReport_top_250_genes_late_high_"
    "refined_int_2026-08-03_extended.csv"
)
late_high_refined_int_ora_extended_df = pd.read_csv(
    late_high_refined_int_ora_extended_path
)
```

```python
# Reactome annotation density is highly uneven, i.e. different
# biological process categories (e.g. DNA repair and autophagy) have
# different numbers of pathway terms
# In an attempt to avoid annotation-density bias, the average of
# -log10(Benjamini) is computed, i.e. the sum of -log10(Benjamini) is
# divided by the number of significantly enriched terms in that category

# As a first step, for each broad biological category, compute the sum
# of the -log10(Benjamini) values
# Simultaneously keep track of the number of significantly enriched
# terms (i.e. terms that satisfy Benjamini < 0.05)
# Additionally, the number of unique genes assigned to the respective
# broad category is also kept track of

# Define the broad biological categories
broad_categories = [
    "Cell cycle & mitosis",
    "DNA replication & repair",
    "Cytoskeleton & Rho GTPase signaling",
    "Membrane trafficking & vesicular transport",
    "Immune response & host–pathogen interactions",
    "Protein & RNA homeostasis",
    "Ubiquitin-mediated protein degradation",
    "Autophagy",
    "Cell signaling",
    "Other"
]

# Lists for storing the sum of -log10(Benjamini)
early_low_unrefined_sum_list = []
early_high_unrefined_sum_list = []
late_low_unrefined_sum_list = []
late_high_unrefined_sum_list = []

early_low_refined_sum_list = []
early_high_refined_sum_list = []
late_low_refined_sum_list = []
late_high_refined_sum_list = []

# Lists for storing the number of significantly enriched terms
early_low_unrefined_n_term_list = []
early_high_unrefined_n_term_list = []
late_low_unrefined_n_term_list = []
late_high_unrefined_n_term_list = []

early_low_refined_n_term_list = []
early_high_refined_n_term_list = []
late_low_refined_n_term_list = []
late_high_refined_n_term_list = []

# Lists for storing the number of unique genes
early_low_unrefined_n_gene_list = []
early_high_unrefined_n_gene_list = []
late_low_unrefined_n_gene_list = []
late_high_unrefined_n_gene_list = []

early_low_refined_n_gene_list = []
early_high_refined_n_gene_list = []
late_low_refined_n_gene_list = []
late_high_refined_n_gene_list = []

for broad_category in broad_categories:
    # As already stated above, an enriched term that satisfies Benjamini
    # < 0.05 for one intensity set may also appear in another intensity
    # set without satisfying the significance condition
    # Thus, Benjamini < 0.05 is applied as additional selection
    # criterion
    # The `Category` selection is applied to err on the side of caution
    # First, deal with unrefined intensities
    (_, early_low_unrefined_b_vals), (_, early_low_unrefined_genes) = (
    early_low_unrefined_int_ora_extended_df.loc[
        (early_low_unrefined_int_ora_extended_df["Category"] == "REACTOME_PATHWAY")
        &
        (early_low_unrefined_int_ora_extended_df["Benjamini"] < 0.05)
        &
        (early_low_unrefined_int_ora_extended_df["Theme"] == broad_category),
        ["minus_log10_Benjamini", "User Ids"]
    ]).items()
    early_low_unrefined_sum_list.append(early_low_unrefined_b_vals.sum())
    early_low_unrefined_n_term_list.append(len(early_low_unrefined_b_vals))
    early_low_unrefined_n_gene_list.append(len(
        early_low_unrefined_genes.str.split(",").explode().unique()
    ))

    (_, early_high_unrefined_b_vals), (_, early_high_unrefined_genes) = (
    early_high_unrefined_int_ora_extended_df.loc[
        (early_high_unrefined_int_ora_extended_df["Category"] == "REACTOME_PATHWAY")
        &
        (early_high_unrefined_int_ora_extended_df["Benjamini"] < 0.05)
        &
        (early_high_unrefined_int_ora_extended_df["Theme"] == broad_category),
        ["minus_log10_Benjamini", "User Ids"]
    ]).items()
    early_high_unrefined_sum_list.append(early_high_unrefined_b_vals.sum())
    early_high_unrefined_n_term_list.append(len(early_high_unrefined_b_vals))
    early_high_unrefined_n_gene_list.append(len(
        early_high_unrefined_genes.str.split(",").explode().unique()
    ))

    (_, late_low_unrefined_b_vals), (_, late_low_unrefined_genes) = (
    late_low_unrefined_int_ora_extended_df.loc[
        (late_low_unrefined_int_ora_extended_df["Category"] == "REACTOME_PATHWAY")
        &
        (late_low_unrefined_int_ora_extended_df["Benjamini"] < 0.05)
        &
        (late_low_unrefined_int_ora_extended_df["Theme"] == broad_category),
        ["minus_log10_Benjamini", "User Ids"]
    ]).items()
    late_low_unrefined_sum_list.append(late_low_unrefined_b_vals.sum())
    late_low_unrefined_n_term_list.append(len(late_low_unrefined_b_vals))
    late_low_unrefined_n_gene_list.append(len(
        late_low_unrefined_genes.str.split(",").explode().unique()
    ))

    (_, late_high_unrefined_b_vals), (_, late_high_unrefined_genes) = (
    late_high_unrefined_int_ora_extended_df.loc[
        (late_high_unrefined_int_ora_extended_df["Category"] == "REACTOME_PATHWAY")
        &
        (late_high_unrefined_int_ora_extended_df["Benjamini"] < 0.05)
        &
        (late_high_unrefined_int_ora_extended_df["Theme"] == broad_category),
        ["minus_log10_Benjamini", "User Ids"]
    ]).items()
    late_high_unrefined_sum_list.append(late_high_unrefined_b_vals.sum())
    late_high_unrefined_n_term_list.append(len(late_high_unrefined_b_vals))
    late_high_unrefined_n_gene_list.append(len(
        late_high_unrefined_genes.str.split(",").explode().unique()
    ))

    # Now, deal with refined intensities
    (_, early_low_refined_b_vals), (_, early_low_refined_genes) = (
    early_low_refined_int_ora_extended_df.loc[
        (early_low_refined_int_ora_extended_df["Category"] == "REACTOME_PATHWAY")
        &
        (early_low_refined_int_ora_extended_df["Benjamini"] < 0.05)
        &
        (early_low_refined_int_ora_extended_df["Theme"] == broad_category),
        ["minus_log10_Benjamini", "User Ids"]
    ]).items()
    early_low_refined_sum_list.append(early_low_refined_b_vals.sum())
    early_low_refined_n_term_list.append(len(early_low_refined_b_vals))
    early_low_refined_n_gene_list.append(len(
        early_low_refined_genes.str.split(",").explode().unique()
    ))

    (_, early_high_refined_b_vals), (_, early_high_refined_genes) = (
    early_high_refined_int_ora_extended_df.loc[
        (early_high_refined_int_ora_extended_df["Category"] == "REACTOME_PATHWAY")
        &
        (early_high_refined_int_ora_extended_df["Benjamini"] < 0.05)
        &
        (early_high_refined_int_ora_extended_df["Theme"] == broad_category),
        ["minus_log10_Benjamini", "User Ids"]
    ]).items()
    early_high_refined_sum_list.append(early_high_refined_b_vals.sum())
    early_high_refined_n_term_list.append(len(early_high_refined_b_vals))
    early_high_refined_n_gene_list.append(len(
        early_high_refined_genes.str.split(",").explode().unique()
    ))

    (_, late_low_refined_b_vals), (_, late_low_refined_genes) = (
    late_low_refined_int_ora_extended_df.loc[
        (late_low_refined_int_ora_extended_df["Category"] == "REACTOME_PATHWAY")
        &
        (late_low_refined_int_ora_extended_df["Benjamini"] < 0.05)
        &
        (late_low_refined_int_ora_extended_df["Theme"] == broad_category),
        ["minus_log10_Benjamini", "User Ids"]
    ]).items()
    late_low_refined_sum_list.append(late_low_refined_b_vals.sum())
    late_low_refined_n_term_list.append(len(late_low_refined_b_vals))
    late_low_refined_n_gene_list.append(len(
        late_low_refined_genes.str.split(",").explode().unique()
    ))

    (_, late_high_refined_b_vals), (_, late_high_refined_genes) = (
    late_high_refined_int_ora_extended_df.loc[
        (late_high_refined_int_ora_extended_df["Category"] == "REACTOME_PATHWAY")
        &
        (late_high_refined_int_ora_extended_df["Benjamini"] < 0.05)
        &
        (late_high_refined_int_ora_extended_df["Theme"] == broad_category),
        ["minus_log10_Benjamini", "User Ids"]
    ]).items()
    late_high_refined_sum_list.append(late_high_refined_b_vals.sum())
    late_high_refined_n_term_list.append(len(late_high_refined_b_vals))
    late_high_refined_n_gene_list.append(len(
        late_high_refined_genes.str.split(",").explode().unique()
    ))
```

```python
# Perform a small sanity check related to the list length
n_broad_categories = len(broad_categories)

assert (
    # Unrefined intensities
    (len(early_low_unrefined_sum_list) == n_broad_categories)
    and
    (len(early_high_unrefined_sum_list) == n_broad_categories)
    and
    (len(late_low_unrefined_sum_list) == n_broad_categories)
    and
    (len(late_high_unrefined_sum_list) == n_broad_categories)
    and
    (len(early_low_unrefined_n_term_list) == n_broad_categories)
    and
    (len(early_high_unrefined_n_term_list) == n_broad_categories)
    and
    (len(late_low_unrefined_n_term_list) == n_broad_categories)
    and
    (len(late_high_unrefined_n_term_list) == n_broad_categories)
    and
    (len(early_low_unrefined_n_gene_list) == n_broad_categories)
    and
    (len(early_high_unrefined_n_gene_list) == n_broad_categories)
    and
    (len(late_low_unrefined_n_gene_list) == n_broad_categories)
    and
    (len(late_high_unrefined_n_gene_list) == n_broad_categories)
    and
    # Refined intensities
    (len(early_low_refined_sum_list) == n_broad_categories)
    and
    (len(early_high_refined_sum_list) == n_broad_categories)
    and
    (len(late_low_refined_sum_list) == n_broad_categories)
    and
    (len(late_high_refined_sum_list) == n_broad_categories)
    and
    (len(early_low_refined_n_term_list) == n_broad_categories)
    and
    (len(early_high_refined_n_term_list) == n_broad_categories)
    and
    (len(late_low_refined_n_term_list) == n_broad_categories)
    and
    (len(late_high_refined_n_term_list) == n_broad_categories)
    and
    (len(early_low_refined_n_gene_list) == n_broad_categories)
    and
    (len(early_high_refined_n_gene_list) == n_broad_categories)
    and
    (len(late_low_refined_n_gene_list) == n_broad_categories)
    and
    (len(late_high_refined_n_gene_list) == n_broad_categories)
), "Something went wrong while populating the lists!"
```

```python
# Compute the average strength, i.e. divide the sum by the number of
# significantly enriched terms assigned to the respective category
# Since some categories may have no assigned terms, special handling of
# division by zero is required
# Also bear in mind that for the condition passed to the `where`
# parameter, the list has to be converted to a NumPy array
# Otherwise, the comparison will simply yield `True`, which is
# broadcasted to the length of the output array

# Unrefined intensities
norm_scores_early_low_unrefined = np.zeros(
    len(broad_categories),
    dtype=float
)
np.divide(
    early_low_unrefined_sum_list,
    early_low_unrefined_n_term_list,
    out=norm_scores_early_low_unrefined,
    where=np.asarray(early_low_unrefined_n_term_list) != 0
)

norm_scores_early_high_unrefined = np.zeros(
    len(broad_categories),
    dtype=float
)
np.divide(
    early_high_unrefined_sum_list,
    early_high_unrefined_n_term_list,
    out=norm_scores_early_high_unrefined,
    where=np.asarray(early_high_unrefined_n_term_list) != 0
)

norm_scores_late_low_unrefined = np.zeros(
    len(broad_categories),
    dtype=float
)
np.divide(
    late_low_unrefined_sum_list,
    late_low_unrefined_n_term_list,
    out=norm_scores_late_low_unrefined,
    where=np.asarray(late_low_unrefined_n_term_list) != 0
)

norm_scores_late_high_unrefined = np.zeros(
    len(broad_categories),
    dtype=float
)
np.divide(
    late_high_unrefined_sum_list,
    late_high_unrefined_n_term_list,
    out=norm_scores_late_high_unrefined,
    where=np.asarray(late_high_unrefined_n_term_list) != 0
)

# Refined intensities
norm_scores_early_low_refined = np.zeros(
    len(broad_categories),
    dtype=float
)
np.divide(
    early_low_refined_sum_list,
    early_low_refined_n_term_list,
    out=norm_scores_early_low_refined,
    where=np.asarray(early_low_refined_n_term_list) != 0
)

norm_scores_early_high_refined = np.zeros(
    len(broad_categories),
    dtype=float
)
np.divide(
    early_high_refined_sum_list,
    early_high_refined_n_term_list,
    out=norm_scores_early_high_refined,
    where=np.asarray(early_high_refined_n_term_list) != 0
)

norm_scores_late_low_refined = np.zeros(
    len(broad_categories),
    dtype=float
)
np.divide(
    late_low_refined_sum_list,
    late_low_refined_n_term_list,
    out=norm_scores_late_low_refined,
    where=np.asarray(late_low_refined_n_term_list) != 0
)

norm_scores_late_high_refined = np.zeros(
    len(broad_categories),
    dtype=float
)
np.divide(
    late_high_refined_sum_list,
    late_high_refined_n_term_list,
    out=norm_scores_late_high_refined,
    where=np.asarray(late_high_refined_n_term_list) != 0
)
```

```python
# With regard to the number of genes, instead of taking the absolute
# value, the fraction of the 250 selected genes is computed
# Computing the fraction for unrefined intensities
early_low_unrefined_gene_frac = (
    np.array(early_low_unrefined_n_gene_list)
    /
    250
)

early_high_unrefined_gene_frac = (
    np.array(early_high_unrefined_n_gene_list)
    /
    250
)

late_low_unrefined_gene_frac = (
    np.array(late_low_unrefined_n_gene_list)
    /
    250
)

late_high_unrefined_gene_frac = (
    np.array(late_high_unrefined_n_gene_list)
    /
    250
)

# Computing the fraction for refined intensities
early_low_refined_gene_frac = (
    np.array(early_low_refined_n_gene_list)
    /
    250
)

early_high_refined_gene_frac = (
    np.array(early_high_refined_n_gene_list)
    /
    250
)

late_low_refined_gene_frac = (
    np.array(late_low_refined_n_gene_list)
    /
    250
)

late_high_refined_gene_frac = (
    np.array(late_high_refined_n_gene_list)
    /
    250
)
```

```python
# Export the results as a TSV file
# Save both the average strength and the sum of -log10(Benjamini)
comparison_average_df = pd.DataFrame(data={
    "category": broad_categories,
    "early_low_unrefined_average": norm_scores_early_low_unrefined,
    "early_low_unrefined_n_terms": early_low_unrefined_n_term_list,
    "early_low_unrefined_genes_frac": early_low_unrefined_gene_frac,
    "early_low_refined_average": norm_scores_early_low_refined,
    "early_low_refined_n_terms": early_low_refined_n_term_list,
    "early_low_refined_genes_frac": early_low_refined_gene_frac,
    "early_high_unrefined_average": norm_scores_early_high_unrefined,
    "early_high_unrefined_n_terms": early_high_unrefined_n_term_list,
    "early_high_unrefined_genes_frac": early_high_unrefined_gene_frac,
    "early_high_refined_average": norm_scores_early_high_refined,
    "early_high_refined_n_terms": early_high_refined_n_term_list,
    "early_high_refined_genes_frac": early_high_refined_gene_frac,
    "late_low_unrefined_average": norm_scores_late_low_unrefined,
    "late_low_unrefined_n_terms": late_low_unrefined_n_term_list,
    "late_low_unrefined_genes_frac": late_low_unrefined_gene_frac,
    "late_low_refined_average": norm_scores_late_low_refined,
    "late_low_refined_n_terms": late_low_refined_n_term_list,
    "late_low_refined_genes_frac": late_low_refined_gene_frac,
    "late_high_unrefined_average": norm_scores_late_high_unrefined,
    "late_high_unrefined_n_terms": late_high_unrefined_n_term_list,
    "late_high_unrefined_genes_frac": late_high_unrefined_gene_frac,
    "late_high_refined_average": norm_scores_late_high_refined,
    "late_high_refined_n_terms": late_high_refined_n_term_list,
    "late_high_refined_genes_frac": late_high_refined_gene_frac
})

comparison_average_df.to_csv(
    "ORA_comparison_average_-log10(Benjamini).tsv",
    sep="\t",
    index=False
)
```

# Converting the Comparison TSV File fron Wide into Long Format

```python
# Convert the comparison DataFrame from wide format into long/tidy
# format
conditions = [
    "early_low_unrefined",
    "early_low_refined",
    "early_high_unrefined",
    "early_high_refined",
    "late_low_unrefined",
    "late_low_refined",
    "late_high_unrefined",
    "late_high_refined"
]

dfs = []

for condition in conditions:
    temp = comparison_average_df[[
        "category",
        f"{condition}_average",
        f"{condition}_n_terms",
        f"{condition}_genes_frac"
    ]].copy()

    temp.columns = [
        "category",
        "average_minus_log10(Benjamini)",
        "n_terms",
        "genes_fraction"
    ]
    temp["ORA"] = condition

    dfs.append(temp)

comparison_average_long_df = pd.concat(dfs, ignore_index=True)

comparison_average_long_df.to_csv(
    "ORA_comparison_average_-log10(Benjamini)_long_format.tsv",
    sep="\t",
    index=False
)
```
