```python
"""
The purpose of this Jupyter notebook is to prepare the
overrepresentation analysis (ORA) via the Functional Annotation Chart
tool of the DAVID Bioinformatics resource.

Specifically, ORA is conducted four times: For both early and late
intensities, genes associated with extraordinarily high and
extraordinarily low intensities are analyzed separately. ORA is
performed for the unrefined intensities as well as for the refined
intensities.
"""
```

```python
import os

import pandas as pd
```

# Load the Data

```python
# Load the early raw and refined intensities
early_ints_path = (
    "unrefined_and_refined_intensities/unrefined_and_refined_early_"
    "intensities.tsv"
)

early_ints_df = pd.read_csv(
    early_ints_path,
    sep="\t"
)

# Load the late raw and refined intensities
late_ints_path = (
    "unrefined_and_refined_intensities/"
    "unrefined_and_refined_late_intensities.tsv"
)

late_ints_df = pd.read_csv(
    late_ints_path,
    sep="\t"
)
```

# Top 250 Genes Selection for Unrefined Intensities

## Sort the Intensities

```python
early_unrefined_ints_df = early_ints_df[["Name", "unrefined_intensity"]].copy()

late_unrefined_ints_df = late_ints_df[["Name", "unrefined_intensity"]].copy()

# Sort by the unrefined intensities in ascending order
early_unrefined_ints_df.sort_values(
    "unrefined_intensity",
    axis=0,
    inplace=True
)

late_unrefined_ints_df.sort_values(
    "unrefined_intensity",
    axis=0,
    inplace=True
)
```

## Select the Top 250 Genes

```python
# For both early and late intensities, the top genes are selected
# The top 250 genes are chosen for high and low intensities each
early_unrefined_high_genes = early_unrefined_ints_df["Name"].iloc[-250:].to_list()
early_unrefined_low_genes = early_unrefined_ints_df["Name"].iloc[:250].to_list()

late_unrefined_high_genes = late_unrefined_ints_df["Name"].iloc[-250:].to_list()
late_unrefined_low_genes = late_unrefined_ints_df["Name"].iloc[:250].to_list()
```

```python
assert (
    (len(early_unrefined_high_genes) == 250)
    and
    (len(early_unrefined_low_genes) == 250)
    and
    (len(late_unrefined_high_genes) == 250)
    and
    (len(late_unrefined_low_genes) == 250)
), "Something went wrong while selecting the top genes!"
```

```python
# Create a subdirectory to store the top genes in
unrefined_top_genes_subdir = "top_250_gene_lists_unrefined_ints"

if not os.path.exists(unrefined_top_genes_subdir):
    os.makedirs(unrefined_top_genes_subdir)
```

```python
# Add a newline character to every gene except the last one
early_unrefined_high_genes_to_file = [
    gene + "\n"
    if i != 249
    else gene
    for i, gene in enumerate(early_unrefined_high_genes)
]

early_unrefined_low_genes_to_file = [
    gene + "\n"
    if i != 249
    else gene
    for i, gene in enumerate(early_unrefined_low_genes)
]

late_unrefined_high_genes_to_file = [
    gene + "\n"
    if i != 249
    else gene
    for i, gene in enumerate(late_unrefined_high_genes)
]

late_unrefined_low_genes_to_file = [
    gene + "\n"
    if i != 249
    else gene
    for i, gene in enumerate(late_unrefined_low_genes)
]
```

```python
# Save the top 250 gene lists to disk
# Bear in mind that in the context of working with files, the `with`
# context manager is preferred as it automatically takes care of closing
# files, even in case of errors/exceptions
with open(
    os.path.join(
        unrefined_top_genes_subdir,
        "top_250_genes_early_high_unrefined_int.txt"
    ),
    "w"
) as f:
    f.writelines(early_unrefined_high_genes_to_file)

with open(
    os.path.join(
        unrefined_top_genes_subdir,
        "top_250_genes_early_low_unrefined_int.txt"
    ),
    "w"
) as f:
    f.writelines(early_unrefined_low_genes_to_file)

with open(
    os.path.join(
        unrefined_top_genes_subdir,
        "top_250_genes_late_high_unrefined_int.txt"
    ),
    "w"
) as f:
    f.writelines(late_unrefined_high_genes_to_file)

with open(
    os.path.join(
        unrefined_top_genes_subdir,
        "top_250_genes_late_low_unrefined_int.txt"
    ),
    "w"
) as f:
    f.writelines(late_unrefined_low_genes_to_file)
```

```python
# For the ORA, not the entire human genome, but all genes interrogated
# in the screen are used as background
# Apart from that, `REACTOME_PATHWAY` as well as `GOTERM_BP_DIRECT` are
# enabled
# EASE Threshold is set to 0.1, Count Threshold to 2
```

# Top 250 Genes Selection for Refined Intensities

## Sort the Intensities

```python
early_refined_ints_df = early_ints_df[["Name", "refined_intensity"]].copy()

late_refined_ints_df = late_ints_df[["Name", "refined_intensity"]].copy()

# Sort by the unrefined intensities in ascending order
early_refined_ints_df.sort_values(
    "refined_intensity",
    axis=0,
    inplace=True
)

late_refined_ints_df.sort_values(
    "refined_intensity",
    axis=0,
    inplace=True
)
```

## Select the Top 250 Genes

```python
# For both early and late intensities, the top genes are selected
# The top 250 genes are chosen for high and low intensities each
early_refined_high_genes = early_refined_ints_df["Name"].iloc[-250:].to_list()
early_refined_low_genes = early_refined_ints_df["Name"].iloc[:250].to_list()

late_refined_high_genes = late_refined_ints_df["Name"].iloc[-250:].to_list()
late_refined_low_genes = late_refined_ints_df["Name"].iloc[:250].to_list()
```

```python
assert (
    (len(early_refined_high_genes) == 250)
    and
    (len(early_refined_low_genes) == 250)
    and
    (len(late_refined_high_genes) == 250)
    and
    (len(late_refined_low_genes) == 250)
), "Something went wrong while selecting the top genes!"
```

```python
# Create a subdirectory to store the top genes in
refined_top_genes_subdir = "top_250_gene_lists_refined_ints"

if not os.path.exists(refined_top_genes_subdir):
    os.makedirs(refined_top_genes_subdir)
```

```python
# Add a newline character to every gene except the last one
early_refined_low_genes_to_file = [
    gene + "\n"
    if i != 249
    else gene
    for i, gene in enumerate(early_refined_low_genes)
]

early_refined_high_genes_to_file = [
    gene + "\n"
    if i != 249
    else gene
    for i, gene in enumerate(early_refined_high_genes)
]

late_refined_low_genes_to_file = [
    gene + "\n"
    if i != 249
    else gene
    for i, gene in enumerate(late_refined_low_genes)
]

late_refined_high_genes_to_file = [
    gene + "\n"
    if i != 249
    else gene
    for i, gene in enumerate(late_refined_high_genes)
]
```

```python
# Save the top 250 gene lists to disk
# Bear in mind that in the context of working with files, the `with`
# context manager is preferred as it automatically takes care of closing
# files, even in case of errors/exceptions
with open(
    os.path.join(
        refined_top_genes_subdir,
        "top_250_genes_early_low_refined_int.txt"
    ),
    "w"
) as f:
    f.writelines(early_refined_low_genes_to_file)

with open(
    os.path.join(
        refined_top_genes_subdir,
        "top_250_genes_early_high_refined_int.txt"
    ),
    "w"
) as f:
    f.writelines(early_refined_high_genes_to_file)

with open(
    os.path.join(
        refined_top_genes_subdir,
        "top_250_genes_late_low_refined_int.txt"
    ),
    "w"
) as f:
    f.writelines(late_refined_low_genes_to_file)

with open(
    os.path.join(
        refined_top_genes_subdir,
        "top_250_genes_late_high_refined_int.txt"
    ),
    "w"
) as f:
    f.writelines(late_refined_high_genes_to_file)
```

```python
# For the ORA, not the entire human genome, but all genes interrogated
# in the screen are used as background
# Apart from that, `REACTOME_PATHWAY` as well as `GOTERM_BP_DIRECT` are
# enabled
# EASE Threshold is set to 0.1, Count Threshold to 2
```

# Investigating the Intersections

```python
# As a sanity check, the intersections between unrefined and refined top
# genes are investigated
# As the refinement alters the composition of the extreme tails, the
# intersection is expected to be small
intersection_early_high = (
    set(early_unrefined_high_genes) & set(early_refined_high_genes)
)

print(
    "Number of genes in intersection of early high genes: "
    f"{len(intersection_early_high)}\n"
)

intersection_early_low = (
    set(early_unrefined_low_genes) & set(early_refined_low_genes)
)

print(
    "Number of genes in intersection of early low genes: "
    f"{len(intersection_early_low)}\n"
)

intersection_late_high = (
    set(late_unrefined_high_genes) & set(late_refined_high_genes)
)

print(
    "Number of genes in intersection of late high genes: "
    f"{len(intersection_late_high)}\n"
)

intersection_late_low = (
    set(late_unrefined_low_genes) & set(late_refined_low_genes)
)

print(
    "Number of genes in intersection of late low genes: "
    f"{len(intersection_late_low)}"
)
```
